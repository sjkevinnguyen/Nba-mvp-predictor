```python
import requests
import pandas as pd
import html5lib

url1 = 'https://www.basketball-reference.com/leagues/NBA_2022_totals.html'
total = pd.read_html(url1, header=0, attrs={'id':'totals_stats'})[0]

url2 = 'https://www.basketball-reference.com/awards/awards_2022.html'
mvp = pd.read_html(url2, header=1, attrs={'id':'mvp'})[0]

url3 = 'https://www.basketball-reference.com/leagues/NBA_2022_advanced.html'
advanced = pd.read_html(url3, header=0, attrs={'id':'advanced'})[0]
```

```python
total_ = total.drop(['Rk', 'Age', 'Team', 'Pos', 'Awards'], axis=1).iloc[:-1]
total_ = total_.drop_duplicates(subset=['Player'], keep='first', ignore_index=True)
total_
```

```python
mvp_ = mvp.drop(['Rank', 'Age', 'Tm'], axis=1)
mvp_
```

```python
advanced_ = advanced.drop(['Rk', 'Age', 'Team', 'Pos', 'Awards'], axis=1).iloc[:-1].drop_duplicates(subset=['Player'], keep='first', ignore_index=True)
advanced_
```

```python
combined = pd.merge(total_, advanced_, on='Player').fillna(0)
pd.set_option('display.max_columns', None)
combined
```

```python
combined2 = pd.merge(combined, mvp_, how='left',on='Player')
combined2
```
